```python
def check_nan_vals(df: pd.DataFrame):
    nan_counts = df.isna().sum()

    total_rows = len(df)

    comparison_df = pd.DataFrame({
        'NaN Count': nan_counts,
        'Total Rows': total_rows,
        'Valid Rows': total_rows - nan_counts
    })

    nan_ratio = df.isna().mean()

    nan_percentage = nan_ratio * 100

    comparison_pct = pd.DataFrame({
        'NaN Count': df.isna().sum(),
        'Total Rows': len(df),
        'Total Rows': len(df),
        'Percentage Missing (%)': nan_percentage.round(2)
    })
    return comparison_df
```

```python
def remove_nan_rows(df: pd.DataFrame, subset: list = None) -> pd.DataFrame:
    """
    Removes all rows containing NaN values and prints row counts before and after.
    """
    initial_rows = len(df)
    
    # Drop rows with NaNs and reset the index cleanly
    cleaned_df = df.dropna(subset=subset).reset_index(drop=True)
    
    final_rows = len(cleaned_df)
    dropped_rows = initial_rows - final_rows
    
    print(f"Rows before: {initial_rows:,}")
    print(f"Rows after:  {final_rows:,}")
    print(f"Dropped:     {dropped_rows:,} ({dropped_rows / initial_rows:.2%} of data)")
    
    return cleaned_df
```

```python
linear_clean = remove_nan_rows(linear_data)
```

```python
with pd.option_context(
    'display.max_rows', None,
    'display.max_columns', None,
    'display.width', None,
    'display.max_colwidth', None
):
    print(check_nan_vals(linear_clean))
```
